```python
import json
import shutil
import os

def copy_coco_images_range(json_path, source_folder, dest_folder, start_index, end_index):
    """
    Reads a COCO JSON, takes images from 'start_index' to 'end_index', 
    and copies them from source_folder to dest_folder.
    """
    
    # 1. Create Destination Folder if it doesn't exist
    if not os.path.exists(dest_folder):
        os.makedirs(dest_folder)
        print(f"Created destination directory: {dest_folder}")

    # 2. Load the COCO JSON file
    print(f"Loading JSON file: {json_path}...")
    try:
        with open(json_path, 'r') as f:
            coco_data = json.load(f)
    except FileNotFoundError:
        print("Error: JSON file not found.")
        return

    # 3. Get the specific range of images
    # Slicing from start_index to end_index (e.g., 500 to 1000)
    images_to_copy = coco_data['images'][start_index:end_index]
    
    total_images = len(coco_data['images'])
    print(f"Total images in JSON: {total_images}")
    print(f"Processing range [{start_index} to {end_index}]")
    print(f"Files to copy in this batch: {len(images_to_copy)}")

    copied_count = 0
    missing_count = 0

    # 4. Loop through and copy
    for img_info in images_to_copy:
        file_name = img_info['file_name']
        
        # Construct full file paths
        src_path = os.path.join(source_folder, file_name)
        dst_path = os.path.join(dest_folder, file_name)

        # Check if file exists before copying
        if os.path.exists(src_path):
            try:
                shutil.copy2(src_path, dst_path)
                copied_count += 1
                
                if copied_count % 50 == 0:
                    print(f"Copied {copied_count} images...")
                    
            except Exception as e:
                print(f"Error copying {file_name}: {e}")
        else:
            print(f"Warning: File not found: {file_name}")
            missing_count += 1

    # 5. Final Summary
    print("-" * 30)
    print("Process Complete.")
    print(f"Batch Range:         {start_index} - {end_index}")
    print(f"Successfully copied: {copied_count}")
    print(f"Missing files:       {missing_count}")
    print(f"Target folder:       {dest_folder}")

# ==========================================
# CONFIGURATION
# ==========================================
if __name__ == "__main__":
    
    # JSON Path
    COCO_JSON_PATH = '/Users/ammaster10/Downloads/annotations 13/500_Relabel.json'
    
    # Correct Source Path (Based on your previous screenshot)
    SOURCE_IMG_DIR = '/Users/ammaster10/Downloads/Test/OCT2017/train/CNV'
    
    # Destination Path (You might want to change this to a new folder like "Extracted_500_1000")
    DEST_IMG_DIR = '/Users/ammaster10/Documents/Github/Year4/CNVresearch/Postprocessing/CNVresearch/Extracted_2'

    # --- SET THE RANGE HERE ---
    START = 500
    END = 1000  

    # Run the function
    copy_coco_images_range(COCO_JSON_PATH, SOURCE_IMG_DIR, DEST_IMG_DIR, START, END)
```
