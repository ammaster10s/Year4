```python
from PyPDF2 import PdfMerger

pdfs = ["/Users/ammaster10/Downloads/CSS434 KB_New/Module 4.pdf", 
        "/Users/ammaster10/Downloads/CSS434 KB_New/Module 5.pdf", 
        "/Users/ammaster10/Downloads/CSS434 KB_New/Module 6.pdf", 
        "/Users/ammaster10/Downloads/CSS434 KB_New/Module 7.pdf"]
merger = PdfMerger()

for pdf in pdfs:
    merger.append(pdf)

merger.write("merged.pdf")
merger.close()

print("PDFs merged into merged.pdf")
```
